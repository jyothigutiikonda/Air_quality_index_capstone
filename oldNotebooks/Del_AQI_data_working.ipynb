```python
import requests
from datetime import datetime, timedelta
```

```python
dates = [datetime(2019, month, 1) for month in range(1, 13)]
for date in dates:
    date_str = date.strftime('%Y%m%d')
    year_str = date.strftime('%Y')
    url = f"https://files.airnowtech.org/airnow/{year_str}/{date_str}/reporting_area.dat"
    print(f"Fetching: {url}")
    response = requests.get(url)
    if response.status_code == 200:
        filename = f"reporting_area_{date_str}.dat"
        with open(filename, 'wb') as f:
            f.write(response.content)
        print(f"Saved: {filename}")
    else:
        print(f"Failed to fetch data for {date_str} - Status code: {response.status_code}")
```
